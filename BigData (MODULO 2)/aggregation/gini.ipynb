```python
import numpy as np
from sklearn.metrics import confusion_matrix

def calculate_gini(y_true, y_pred):
    # escludiamo il rumore del DBSCAN
    valid_mask = (y_pred != -1) # restituisce gli indici associati ai valori validi
    if valid_mask.sum() == 0:
        return 1.0 # se è tutto rumore restituiamo il massimo, Gini impuro
    y_t = y_true[valid_mask]
    y_p = y_pred[valid_mask]

    # calcolo della matrice di confusione
    cm = confusion_matrix(y_t,y_p)

    # M_j è la dimensione totale di ciascun cluster j (somma sulle colonne)
    M_j = cm.sum(axis=0)

    g_j_list = []

    # Calcolo del gini locale G_j per ciascun cluster j

    for j in range(cm.shape[1]):
        # Controllo anti-divisione per zero: procediamo solo se il cluster 
        # ha punti (M_j>0)
        if M_j[j] > 0:
            sum_sq = np.sum((cm[:,j] / M_j[j]) ** 2)
            g_j = 1 - sum_sq
            g_j_list.append((g_j, M_j[j]))

    if not g_j_list:
        return 1.0

    # Calcolo del Gini medio ponderato globale (G_average)
    numerator = sum(g * m for g,m in g_j_list)
    denominator = sum(m for _,m in g_j_list)

    if denominator == 0:
        return 1.0

    return numerator/denominator
```
