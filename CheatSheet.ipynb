```python
"""
CHEAT SHEET - PANDAS, NUMPY, SKLEARN
Argomenti: Selezione, Pulizia, Discretizzazione, PCA, SVD
"""

import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# ==========================================
# 1. PANDAS: Caricamento e Selezione
# ==========================================
# Caricare dati senza intestazione, fornendo noi i nomi
labels = ['col1', 'col2'] + list(range(10))
df = pd.read_csv('file.csv', names=labels)

# Ispezione
df.head()          # Prime 5 righe
df.shape           # Dimensioni (righe, colonne)
df.describe()      # Statistiche (media, min, max, std)

# Selezione Dati
df.loc['riga_A', 'col_B']  # Usa le ETICHETTE (nomi)
df.iloc[0, 5]              # Usa gli INDICI NUMERICI (posizioni)
df.at['riga_A', 'col_B']   # Come loc, ma per un SINGOLO valore (veloce)
df.iat[0, 5]               # Come iloc, ma per un SINGOLO valore (veloce)

# Slicing (Tagliare il dataset)
# Tutte le righe, colonne dalla terza alla penultima
subset = df.iloc[:, 2:-1] 

# Convertire DataFrame in matrice NumPy (fondamentale per Sklearn/Math)
matrice = df.to_numpy()

# ==========================================
# 2. PULIZIA DATI (Data Cleaning)
# ==========================================
# Metodo Veloce (Vettoriale)
df.replace('?', np.nan, inplace=True) # Sostituisce valore specifico
df.fillna(0, inplace=True)            # Riempie i buchi con 0 (o df.mean())
df = df.astype(float)                 # Converte tutto in numeri decimali

# Metodo Lento (Manuale - Didattico)
# Richiede doppio ciclo for + df.iat[r, c]

# ==========================================
# 3. DISCRETIZZAZIONE (Binning)
# ==========================================
# Equi-Width (Intervalli di larghezza uguale) -> Sensibile agli outlier
bins_width = pd.cut(df['colonna'], bins=10, labels=False)

# Equi-Depth (Stessa frequenza/quantità per bin) -> Istogramma piatto
# duplicates='drop' serve se ci sono troppi valori identici
bins_depth = pd.qcut(df['colonna'], q=10, labels=False, duplicates='drop')

# ==========================================
# 4. PCA (Scikit-Learn)
# ==========================================
# n_components=0.8 -> Mantieni l'80% della varianza
# n_components=5   -> Mantieni esattamente 5 colonne
pca = PCA(n_components=0.8, svd_solver='full')

# Fit (Impara) + Transform (Comprime)
d_new = pca.fit(matrice).transform(matrice)

# Attributi utili
pca.n_components_            # Quante componenti sono rimaste?
pca.explained_variance_ratio_ # Quanta info contiene ogni componente?
pca.singular_values_          # I valori singolari (importanti per confronto SVD)

# ==========================================
# 5. SVD & ALGEBRA LINEARE (NumPy)
# ==========================================
# SVD Pura (compute_uv=False restituisce solo Sigma, risparmia memoria)
sigma = np.linalg.svd(matrice, compute_uv=False)

# Centratura manuale (Fondamentale per far coincidere SVD e PCA)
media = matrice.mean(axis=0)  # Media lungo le colonne
matrice_centrata = matrice - media

# Eigen-decomposition (Autovalori) su Matrice di Covarianza
C = np.matmul(matrice_centrata.T, matrice_centrata) # D^T * D
autovalori, autovettori = np.linalg.eig(C)

# Relazione matematica chiave
# Radice Quadrata Autovalori Matrice Covarianza ~= Valori Singolari SVD
np.sqrt(np.sort(autovalori)[::-1])
```
