```python
import numpy as np

# Creiamo una matrice A (3 documenti, 2 parole)
A = np.array([[1, 2], 
              [2, 1], 
              [10, 8]]) # Il terzo documento è molto diverso

# Applichiamo la SVD
U, s, VT = np.linalg.svd(A)

# S è restituito come un vettore per efficienza, lo trasformiamo in matrice diagonale
Sigma = np.zeros(A.shape)
np.fill_diagonal(Sigma, s)

print("Matrice U (Relazione Documenti-Concetti):\n", U)
print("\nValori Singolari (Importanza):\n", s)
print("\nMatrice VT (Relazione Parole-Concetti/Vettori Base):\n", VT)

# Verifica: U * Sigma * VT deve ridare A
A_ricostruita = U @ Sigma @ VT
print("\nMatrice Ricostruita:\n", A_ricostruita)
```
