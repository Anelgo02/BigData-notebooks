**SAX (Symbolic Aggregate approXimation)**

La SAX trasforma una sequenza temporale in una sequenza discreta, il processo si divide in due step:

- La serie viene divisa in finestre di lunghezza w e viene calcolato il valore medio di ogni finestra.

- Discretizziamo ciascun intervallo in un piccolo numero di intervalli e ciascuno di essi verra' mappato ad un valore discreto.

```python
import numpy as np

# 1. Serie temporale di esempio
ts = np.array([1.1, 1.5, 0.8, 2.2, 0.5, -0.2, -1.0, -0.5])

# 2. Normalizzazione
ts_norm = (ts - np.mean(ts)) / np.std(ts)

# 3. PAA (Piecewise Aggregate Approximation), dividiamo in finestre di 2 colonne e calcoliamo la media,
# il vettore risultante sara' => [3, 3, 9, 3]

paa = ts_norm.reshape(-1,2).mean(axis=1)


# Funzione di discretizzazione, definiamo dei breakpoint arbitrari
def map_to_sax(value):
    if value < -0.43: return 'a'
    if value < 0.43: return 'b'
    return 'c'

sax_string = "".join([map_to_sax(v) for v in paa])
print(f"Stringa SAX risultante: {sax_string}")    
```
